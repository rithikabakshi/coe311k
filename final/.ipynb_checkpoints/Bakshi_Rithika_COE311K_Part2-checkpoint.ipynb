### Section 1: Introduction & System Selection
The system I selected is cooling with forced convection, which is governed by the equation $\frac{dT}{dt} = -h(T-T_{\infty}(t))$

The parameters being used are:

$h = -1000s^{-1} $, h is how effective the heat is transferred, this is very high

$ T_{\infty}(t) = 300 + 50sin(0.01t)K $, this is the ambient temperature, which slowly varies over time

$T(0) = 500K$, initial temperature is 500 Kelvin

This system is important in practice for devices like batteries, where heat is generated and has to be dissipated. For systems with a high h-value, active cooling is being used.

### Section 2: Demonstration of Stiffness

```python
import numpy as np
import matplotlib.pyplot as plt
def euler(step):
    # Define parameters
    h = step  # step size
    x_start, x_end = 0, 5
    num_steps = int((x_end - x_start) / h)

    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)


    # Initial condition
    y_values[0] = 0  # y(0) = 0

    # Implement Euler's method

    for i in range(num_steps):

        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
        
    return x_values, y_values

def f(t,T):
    return -1000*T + 1000*(300+50*np.sin((0.01*t)))
```

```python
plt.figure(figsize=(12, 8))

stepsizes = [0.01,0.00001]
for i in stepsizes:  
    x, y = euler(i)
    plt.plot(x,y, label = "forward euler")



plt.xlabel('x')
plt.ylabel('y')
# plt.title("dy/dx = -2y using Euler's Method")

# xref = np.arange(0,30,0.1)
# yref = 3 - 0.998*np.exp(-1000*xref) - 2.002*np.exp((-1)*xref)
# plt.plot(xref,yref, label = "Analytical Solution")

plt.legend()
plt.show()
```

It is pretty clear that stepsizes 0.1 and 0.01 completely break, but the smaller forward Euler approximations are a relatively good match, even though they are extremely computationally expensive. 
The stiffness ratio is 1000/0.01 = 100,000. 
Implicit methods are essential for stiff equations because they stay stable even with large time steps, so your simulation won't "blow up" or crash. This allows you to skip tiny, unnecessary calculations and focus on the main trends of the system, making the whole process much faster and more reliable.

### Section 3: Mathematical Setup

Residual Function:  $g(T_{n+1}) = T_{n+1} - T_n - \Delta t \left[ -1000T_{n+1} + 1000(300 + 50 \sin(0.01t_{n+1})) \right]$

This is derived by moving $T_{n+1}$ to the other side of the equation, solving for when this equation equals 0.

Analytical Derivative: $g'(T_{n+1}) = 1 + 1000\Delta t$

The derivative of the residual function with respect to $T_{n+1}$

Iteration Formula: $y_{n+1} = y_{n+1} - \frac{g(y_{n+1})}{g'(y_{n+1})}$

### Section 4: Implementation

```python
def newton_raphson(g, dg, x0, tol=1e-6, max_iter=50):
    x = x0
    for i in range(max_iter):
        x_new = x - g(x) / dg(x)
        if abs(x_new - x) < tol:
            return x_new, i + 1
        x = x_new
    return x, max_iter

def f(t, T):
    h_coeff = 1000
    T_inf = 300 + 50 * np.sin(0.01 * t)
    return -h_coeff * (T - T_inf)

def backward_euler_stiff(f, y0, t0, tf, dt):
    t_values = np.arange(t0, tf + dt, dt)
    y_values = np.zeros_like(t_values)
    y_values[0] = y0
    
    h_val = 1000
    dg = lambda y_next: 1 + h_val * dt

    for i in range(1, len(t_values)):
        t_next = t_values[i]
        y_prev = y_values[i - 1]
        
        # Define g for this specific time step
        g_step = lambda y_next: y_next - y_prev - dt * f(t_next, y_next)
        
        # Solve for y_next using your NR function
        # We take only the first return value [0] which is the root
        y_next, _ = newton_raphson(g_step, dg, x0=y_prev)
        
        y_values[i] = y_next
        
    return t_values, y_values

T0 = 500
t_start = 0
t_end = 10 # Adjust as needed to see the sine wave
dt = 0.1   # We can use a much larger dt than Forward Euler

t_stiff, T_stiff = backward_euler_stiff(f, T0, t_start, t_end, dt)

def forward_euler(step):
    # Define parameters
    h = step  # step size
    x_start, x_end = 0, 30
    num_steps = int((x_end - x_start) / h)

    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)


    # Initial condition
    y_values[0] = 0  # y(0) = 0

    # Implement Euler's method

    for i in range(num_steps):

        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
        
    return x_values, y_values
```

### Section 5: Results & Analysis

```python
dt_unstable = 0.0021 
dt_stable_large = 0.1

t1, T_fe = forward_euler(dt_unstable)
t2, T_be_unstable_dt = backward_euler_stiff(f, T0, t_start, t_end, dt_unstable)
t3, T_be_large_dt = backward_euler_stiff(f, T0, t_start, t_end, dt_stable_large)


plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(t1, T_fe, 'r--', label=f'Forward Euler (dt={dt_unstable})')
plt.plot(t2, T_be_unstable_dt, 'b-', label=f'Backward Euler (dt={dt_unstable})')
plt.ylim(200, 600)
plt.title("Stability Comparison (dt > 2/h)")
plt.xlabel("Time (s)")
plt.ylabel("Temperature (K)")
plt.legend()

# Plot 2: Efficiency (Large Steps)
plt.subplot(1, 2, 2)
plt.plot(t3, T_be_large_dt, 'g-o', markersize=4, label=f'Backward Euler (dt={dt_stable_large})')
plt.title("Backward Euler (Large Step Size)")
plt.xlabel("Time (s)")
plt.ylabel("Temperature (K)")
plt.legend()

plt.show()
```

### Section 6: Performance Analysis
Even though the math for the Backward Euler (implicit) method is more complex because it requires solving a root-finding problem at every step, it actually saves a significant amount of time in the long run. With Forward Euler, the user is forced to use a tiny step size to keep the code from crashing or diverging. For this specific stiff problem, Backward Euler allows for a step size that is 50x to 100x larger, meaning the computer performs far fewer total operations to reach the same result.

- Forward Euler: Simple to implement but highly unstable for stiff problems, requiring an extremely small step size ($dt < 0.002$) which makes long simulations very slow
- Backward Euler: More complex to set up due to the Newton-Raphson requirement, but it is unconditionally stable, allowing for much larger step sizes and much faster total computation times.

If a large step size is used with Backward Euler, the code runs almost instantly, but some detail might be lost. However, since this system is so "stiff," the object’s temperature follows the ambient temperature almost perfectly anyway. This means a larger step size can be used without sacrificing much accuracy.

When dealing with a "stiff" system where variables change rapidly (like $h=1000$), it makes more sense to use an implicit method. While it takes a bit more effort to set up the Newton-Raphson logic, it prevents the simulation from taking forever to finish and ensures the results don't "blow up" to infinity.

### Section 7: Conclusions

The main takeaway from this project is how much a "stiff" equation changes the approach needed to solve a problem. Because the cooling coefficient was so high ($h = 1000$), the object's temperature changed too fast for a standard explicit solver to handle without crashing. By switching to the Backward Euler method and using Newton-Raphson, the simulation remained stable regardless of the step size. Even with the air temperature oscillating as a sine wave, the implicit solver stayed accurate while using much larger steps than the Forward Euler method

Explicit methods like Forward Euler are useful for simple problems because the code is straightforward and does not require loops. However, for stiff systems where different parts move at  different speeds, implicit methods are necessary. Generally, if a tiny step size is required just to stop a graph from "blowing up" or going to infinity, it's a sign that an implicit solver is needed.

The most difficult part was setting up the math for the Newton-Raphson iterations. It takes more effort to derive the residual function and its derivative than it does to plug numbers into a basic formula. This project showed that numerical stability is important, since  a correct formula is not useful if the simulation takes too long to run.
