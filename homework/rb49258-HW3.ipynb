```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def taylorSeries(x, numTerms):
    sum = 0
    for i in range(0,numTerms):
        term = ((-1)**i)*(x**(2*i+1))/np.math.factorial((2*i+1))
        # print(i,": ",term)
        sum += term
    return sum
```

```python
print("testing with sin(1)")
for i in range(1,5):
    test = taylorSeries(1,i)
    print(i,": ",test)
    
```

```python
xAxis = np.arange(0,10,0.1)
# print(xAxis)
oneTerm = taylorSeries(xAxis,1)
# print(oneTerm)
twoTerm = taylorSeries(xAxis,2)
# print(twoTerm)
threeTerm = taylorSeries(xAxis,3)
# print(threeTerm)
fourTerm = taylorSeries(xAxis,4)
print(fourTerm)
```

```python

plt.plot(xAxis, oneTerm, 'r', label = "one term")
plt.plot(xAxis, twoTerm, 'g', label = "two term")
plt.plot(xAxis, threeTerm, 'b', label = "three term")
plt.plot(xAxis, fourTerm, 'm', label = "four term")
plt.plot(xAxis, np.sin(xAxis), 'k--', label="actual sin")
plt.ylim(-1.5, 1.5)
plt.legend()
plt.show()
```

```python
oneTermErr = abs(oneTerm-np.sin(xAxis))
twoTermErr = abs(twoTerm-np.sin(xAxis))
threeTermErr = abs(threeTerm-np.sin(xAxis))
fourTermErr = abs(fourTerm-np.sin(xAxis))
```

```python
plt.plot(xAxis, oneTermErr, 'r', label = "one term error")
plt.plot(xAxis, twoTermErr, 'g', label = "two term error")
plt.plot(xAxis, threeTermErr, 'b', label = "three term error")
plt.plot(xAxis, fourTermErr, 'm', label = "four term error")
plt.ylim(0, 20)
plt.legend()
plt.show()
```

As more terms are added to the series, the error increases later. Because the one term error is (1-sin(x)), it oscillates back and forth as its slope changes. The error changes are unpredictible on a higher scale and do not increase uniformly. 

```python
xAxisExp = np.arange(0,20,0.5)

oneTerm = taylorSeries(xAxisExp,1)
twoTerm = taylorSeries(xAxisExp,2)
threeTerm = taylorSeries(xAxisExp,3)
eightTerm = taylorSeries(xAxisExp,8)


oneTermErr = abs(oneTerm-np.sin(xAxisExp))
twoTermErr = abs(twoTerm-np.sin(xAxisExp))
threeTermErr = abs(threeTerm-np.sin(xAxisExp))
eightTermErr = abs(eightTerm-np.sin(xAxisExp))
plt.plot(xAxisExp, oneTermErr, 'r', label = "one term error")
plt.plot(xAxisExp, twoTermErr, 'g', label = "two term error")
plt.plot(xAxisExp, threeTermErr, 'b', label = "three term error")
plt.plot(xAxisExp, eightTermErr, 'm', label = "eight term error")
plt.ylim(0, 1000)
plt.legend()
plt.show()
```

For larger values of x, the graph shows the same trend, with the more expanded polynomials creating error at a faster rate, but also taking longer to start creating the error.
