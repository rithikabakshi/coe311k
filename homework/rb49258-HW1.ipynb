```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
def approximatePi(numSamples):
    counter = 0
    
    for s in range(0,numSamples):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        if(x*x + y*y<1):
            counter+=1
    piEst = 4*(counter/numSamples)
    return piEst
```

```python
arr = np.array([])
for i in range(10,10000,20):
    errorDiff = np.pi-approximatePi(i)
    arr = np.append(arr,errorDiff)
np.round(arr,15)
```

```python
plt.hist(arr,30)
plt.show
```
