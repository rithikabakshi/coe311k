```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x, y):
    return -2 * y * np.cos(x)
    
def euler(step):
    # Define parameters
    h = step  # step size
    x_start, x_end = 0, 10*np.pi
    num_steps = int((x_end - x_start) / h)

    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)


    # Initial condition
    y_values[0] = 1  # y(0) = 1

    # Implement Euler's method

    for i in range(num_steps):

        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
        
    return x_values, y_values
```

```python
stepsize = np.arange(0,1,0.05)
stepsize[0] = 0.01
y_average = []
y_closest15 = []
max_errors = []

plt.figure(figsize=(12, 8))

xref, yref = euler(0.001)

yref_average = np.trapz(yref, xref) / (xref[-1] - xref[0])
min15 = np.argmin(np.abs(xref - 15))
yref15 = yref[min15]

for i in stepsize:
    x_values, y_values = euler(i)
    
    yavg = np.trapz(y_values, x_values) / (x_values[-1] - x_values[0])
    y_average.append(yavg)
    
    min15 = np.argmin(np.abs(x_values - 15))
    yclosest15 = y_values[min15]
    y_closest15.append(yclosest15)

    
#     true_vals = np.exp(-2 * np.sin(x_values))
#     pointwise_error = np.abs(y_values - true_vals)
#     max_error = np.max(pointwise_error)
#     max_errors.append(max_error)
    
    y_cur_interp = np.interp(xref, x_values, y_values)
    max_error = np.max(np.abs(y_cur_interp - yref))
    max_errors.append(max_error)

    plt.plot(x_values, y_values, label=f"Step size = {i:.2f}")

plt.xlabel('x')

plt.ylabel('y')

plt.title("Approximation of dy/dx = -2ycos(x) using Euler's Method")

plt.plot(xref,yref,'r--',label = "reference")

plt.legend()

plt.show()
```

```python
# average the y values of for each run of h
error = abs(np.array(y_average) - yref_average)

plt.scatter(stepsize, error)

plt.xlabel('h')

plt.ylabel('Average Error')

plt.title("Average Error Between Runs")

plt.show()

error = abs(np.array(y_closest15) - yref15)

plt.scatter(stepsize, error)

plt.xlabel('h')

plt.ylabel('Local Error')

plt.title("Local Error Between Runs at x~15")

plt.show()
```
