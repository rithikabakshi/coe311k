```python
import numpy as np
import matplotlib.pyplot as plt

def euler(step):
    # Define parameters
    h = step  # step size
    x_start, x_end = 0, 30
    num_steps = int((x_end - x_start) / h)

    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)


    # Initial condition
    y_values[0] = 0  # y(0) = 0

    # Implement Euler's method

    for i in range(num_steps):

        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
        
    return x_values, y_values

def f(t,y):
    return -1000*y + 3000 - 2000*np.exp((-1)*t)
```

```python
plt.figure(figsize=(12, 8))

stepsizes = [0.1,.01, .001, .00001, .0000001]
for i in stepsizes:  
    x, y = euler(i)
    plt.plot(x,y, label = "forward euler")



plt.xlabel('x')
plt.ylabel('y')
# plt.title("dy/dx = -2y using Euler's Method")

xref = np.arange(0,30,0.1)
yref = 3 - 0.998*np.exp(-1000*xref) - 2.002*np.exp((-1)*xref)
plt.plot(xref,yref, label = "Analytical Solution")

plt.legend()
plt.show()
```
