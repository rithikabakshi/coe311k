```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng()

a3 = rng.random(size=(3, 3))
b3 = rng.random(size=(3, 3))

a6 = rng.random(size=(6, 6))
b6 = rng.random(size=(6, 6))

a12 = rng.random(size=(12, 12))
b12 = rng.random(size=(12, 12))

a24 = rng.random(size=(24, 24))
b24 = rng.random(size=(24, 24))

a48 = rng.random(size=(48, 48))
b48 = rng.random(size=(48, 48))

a96 = rng.random(size=(96, 96))
b96 = rng.random(size=(96, 96))

a192 = rng.random(size=(192, 192))
b192 = rng.random(size=(192, 192))

a384 = rng.random(size=(384, 384))
b384 = rng.random(size=(384, 384))
```

```python
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C
```

```python
%%time
naive_matrix_multiplication(a3,b3)
t3 = 143*10**-6
```

```python
%%time
naive_matrix_multiplication(a6,b6)
t6 = 525*10**-6
```
