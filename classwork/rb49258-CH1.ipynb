```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi, 50)
```

```python
plt.subplot(2, 2, 1) # (row, column, active area)
plt.plot(x, np.sin(x), 'r')
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x), 'g')
plt.subplot(2,2, 3)
plt.plot(x, np.arcsin(x)*2, 'b')
plt.subplot(2,2, 4)
plt.scatter(x,np.tan(x))
plt.show()
```
