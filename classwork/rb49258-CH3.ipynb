```python
import numpy as np
import random
import matplotlib.pyplot as plt
import math
```

```python
def approximatePi(numSamples):
    counter = 0
    
    for s in range(0,numSamples):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        if(x*x + y*y<1):
            counter+=1
    piEst = 4*(counter/numSamples)
    return piEst
```

```python
trueerr = []
relerr = []
xaxis = np.arange(0,100)
for x in range(0,100):
    approx = approximatePi(1000)
    
    trueerr = np.append(trueerr, abs(approx - np.pi))
    relerr = np.append(relerr, abs(approx - np.pi) / np.pi)  
    
plt.subplot(2,2,1)
plt.hist(trueerr, 10)
plt.title('True Error Histogram')

plt.subplot(2,2,2)
plt.hist(relerr, 10)
plt.title('Relative Error Histogram')

plt.subplot(2,2,3)
plt.plot(xaxis, trueerr)
plt.title('True Error Plot')

plt.subplot(2,2,4)
plt.plot(xaxis, relerr)
plt.title('Relative Error Plot')
```
