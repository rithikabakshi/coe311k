```python
import numpy as np
import matplotlib.pyplot as plt

# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y
    
def euler(step):
    # Define parameters
    h = step  # step size
    x_start, x_end = 0, 2
    num_steps = int((x_end - x_start) / h)

    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)


    # Initial condition
    y_values[0] = 1  # y(0) = 1

    # Implement Euler's method

    for i in range(num_steps):

        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
        
    return x_values, y_values
```

```python
def exact(x):
    return np.exp(-2 * x)

def compute_errors(x_vals, y_approx):
    y_true = exact(x_vals)

    local_error = np.abs(y_approx - y_true)         
    global_error = np.max(local_error)              

    return local_error, global_error
```

```python
plt.figure(figsize=(12, 8))
xref, yref = euler(0.01)
for i in np.arange(0.01, 1, 0.05):
    x_values, y_values = euler(i)
    
    local_err, global_err = compute_errors(x_values, y_values)

    print(f"\nStep size h = {i}")
    print(f"Global error (max): {global_err:.6f}")

    plt.plot(x_values, local_err, label=f"h = {i}")

plt.xlabel('x')

plt.ylabel('y')

plt.title("Error of dy/dx = -2y using Euler's Method")

plt.legend()

plt.show()
    
```
