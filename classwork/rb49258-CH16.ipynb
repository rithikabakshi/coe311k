```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters

def eulers(k,h,xend):
    
    k = k       # Constant in the equation (e.g., spring constant)
    h = h     # Step size
    x_start = 0    # Start of the interval
    x_end = xend     # End of the interval
    y_0 = 1.0      # Initial condition for y
    v_0 = 0.0      # Initial condition for dy/dx

    # Create arrays to store x, y, and v values

    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    # Set initial conditions

    y_values[0] = y_0
    v_values[0] = v_0



    # Euler's Method loop

    for i in range(1, len(x_values)):
        # Current values
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]    

        # Update equations
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)
        
    return x_values,y_values,v_values
```

```python
# Plot the results
plt.figure(figsize=(10, 5))
stepsizes = [.01, .02, .03, .05, .1, .2, .3, .5]
for i in stepsizes:
    x_values, y_values, v_values = eulers(1.0, i, 10)
    plt.plot(x_values, y_values, label=f"h = {i}")
    plt.plot(x_values, v_values, label=f"h = {i}", linestyle='--')


plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend()
plt.grid(True)

plt.show()
```

Even a minor change in step size such as h = 0.01 to h = 0.02 causes the wave to increase in its amplitude, showing that the Euler's method solution is unstable. As the step size grows, the solution diverges faster.

```python
# Plot the results
plt.figure(figsize=(10, 5))
x_values, y_values, v_values = eulers(10.0, 0.01, 20)
plt.plot(x_values, y_values, label=f"k = {10}")
plt.plot(x_values, v_values, label=f"k = {10}", linestyle='--')
x_values, y_values, v_values = eulers(1.0, 0.01, 20)
plt.plot(x_values, y_values, label=f"k = {1}")
plt.plot(x_values, v_values, label=f"k = {1}", linestyle='--')



plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend()
plt.grid(True)

plt.show()
```

The value of k also affects the stability of the solution. As seen in the graph, the higher k value causes the amplitudes to grow as x increases. This is because increasing the k value makes the peaks of the wave higher, which Euler's method is unable to keep up with, causing it to produce an unstable solution.

```python
plt.figure(figsize=(10, 5))
x_values, y_values, v_values = eulers(10.0, 0.01, 20)
plt.plot(x_values, y_values, label="position")
plt.plot(x_values, v_values, label="velocity", linestyle='--')

x_values, y_values, v_values = eulers(10.0, 0.001, 20)
plt.plot(x_values, y_values, label="position small step")
plt.plot(x_values, v_values, label="velocity small step", linestyle='--')



plt.xlabel('x')
plt.ylabel('y and v')
plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
plt.legend()
plt.grid(True)

plt.show()
```

When the h is set to a small value like 0.001 (seen in green and red), the solution produced by Euler's method remains stable. With a larger h like 0.01 (seen in blue and orange), the oscillations keep growing, indicating an unstable solution.
