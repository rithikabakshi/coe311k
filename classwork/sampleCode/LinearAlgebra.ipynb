determinant:
A = [[a,b],[c,d]]
det(A) = ad-bc
- abs val shows how much itll stretch/shrink
- 0 shows itll collapse into a line
- negative det flips orientation

- if system = 0, no unique sol exists

near-zero determinant: unstable


```python
import numpy as np
A = np.array([[2,1],[1,1]])
B = np.array([[2,4],[1,2]])
print(np.linalg.det(A)) # not flipped, unique sol exists, stable matrix
print(np.linalg.det(B)) # multiples of each other make det = 0, unstable matrix
```

```python
Ainv = np.linalg.inv(A)
# Binv = np.linalg.inv(B)

print(Ainv)
# print(Binv)
```

if two line system crosses at a large angle, a smalll shift won't change solution that much
if the angle is small, that can cause really large horizontal/vert shift
kappa: how sensitive the matrix is to change
- well conditioned: ~1
- ill-conditioned: very large

consequences
- rounding errors
- difficulty solving w jacobi or gauss-seidel

```python
A = np.array([[1, 	1], [1, 1.0001]])
```

```python
print(A)
kappa = np.linalg.cond(A,2)
print(kappa)
```
