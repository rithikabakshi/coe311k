multiple lines intersecting at different spots (more equations than unknown)
least squares method: find best fitting solution to a system of equations
samllest amount of residual error (square of vertical distance)

why square? makes error positive, magnifies larger residuals

how to minimize
- take partial derivative of residual (S(m,c)) with respect to m and c
- set them equal to 0
- gives a system of linear equations known as the normal equations
- solve for values of m and c that minimizes the error

```python
import numpy as np
# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])
# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

print("X: ",X)
print("Y: ",Y)
print("A: ",A)
```

```python
# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}"
```
