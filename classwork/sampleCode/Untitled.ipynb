```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def leibniz(terms):
    sum = 0
    for n in range(0,terms):
        sum+=((-1)**n)/(2*n+1)
        if(n%100==0):
            print("rel err: ",(abs(np.pi-sum*4)/np.pi))
    return sum
```

```python
test = leibniz(5000)
print(test*4)
relErr = np.pi-test*4
```
