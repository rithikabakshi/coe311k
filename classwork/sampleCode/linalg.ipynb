### eigenvalues and eigenvectors
for a square matrix A, there exists a lambda and a vector v such that Av = lambda*v
- eigenvectors are directions tha tdon't change when a matrix acts on them
- eigenvalues tell you how much stretch those directions have (negative = decay, positive = growth, complex = oscillations






```python
A = np.array([[2,1],[1,2]])
eigenvalue, eigenvectors = np.linalg.eig(A)
```

```python
print(eigenvalue, eigenvectors)
```

#### singular value decomposition
Singular Value Decomposition (SVD) is a factorization of a matrix into three
matrices:
A=UΣV T
Where:

● U and V are orthogonal matrices,

● Σ is a diagonal matrix of singular values

```python
# Define a matrix
A = np.array([[1, 2], [3, 4], [5, 6]])
# Perform SVD
U, sigma, Vt = np.linalg.svd(A)
print("U matrix:\n", U)
print("Sigma values:", sigma)
print("V^T matrix:\n", Vt)
```
