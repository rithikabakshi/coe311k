### Introduction

The dataset used in this analysis consists of quarterly U.S. GDP growth rates, including the sharp decline during the COVID-19 period. This data comes from the U.S. Bureau of Economic Analysis (BEA) and reflects how the economy has changed over time from 2010 to 2023. To analyze the data, three different methods are applied: cubic spline interpolation to estimate values between known quarters, polynomial fitting to model the data with a global curve, and a least squares linear model to identify an overall trend after removing the COVID quarters. Each method highlights different aspects of the data and helps compare how well they handle variability and extreme values.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# create dataset, plot for checking

x = np.array([1,5,9,13,17,19,22,25,27,28,29,33,37,41,42,43,45,49,54,56])

y = np.array([1.7,0.1,2.3,2.7,1.7,5.0,3.0,1.5,3.5,1.8,1.3,2.5,3.1,-5.1,-28.1,33.8,6.3,-1.6,2.4,3.3])

plt.xlabel("Quarter")
plt.ylabel("GDP Growth (%)")
plt.title("Quarter vs GDP Growth, Given Points")

plt.scatter(x,y, label = "Given Points")
plt.legend()
plt.show()
```

### Part A - Cubic Spline Interpolation

```python
def setup_tridiagonal_matrix(x, y):
    n = len(x)
    h = np.diff(x) # Step sizes between data points
    # Initialize the tridiagonal matrix coefficients
    A = np.zeros(n - 2) # Subdiagonal
    B = np.zeros(n - 2) # Main diagonal
    C = np.zeros(n - 2) # Superdiagonal
    D = np.zeros(n - 2) # Right-hand side vector
    # Populate the tridiagonal matrix coefficients
    for i in range(1, n - 1):
        A[i - 1] = h[i-1] # Subdiagonal
        B[i - 1] = 2*(h[i-1]+h[i]) # Main diagonal
        C[i - 1] = h[i] # Superdiagonal
        D[i - 1] = 6*(((y[i+1]-y[i])/h[i])-((y[i]-y[i-1])/h[i-1])) # Right-hand side vector
    return A, B, C, D
```

```python
def thomas_algorithm(a, b, c, d):
    """
    Solve the tridiagonal system using the Thomas algorithm.
    a: sub-diagonal elements (length n-1)
    b: main diagonal elements (length n)
    c: super-diagonal elements (length n-1)
    d: right-hand side (length n)
    Returns: solution vector x of length n
    """
    n = len(b)
    # Forward elimination
    for i in range(1, n):
        w = a[i-1] / b[i-1]
        b[i] = b[i] - w * c[i-1]
        d[i] = d[i] - w * d[i-1]
        # Back substitution
    M = np.zeros(n)
    M[-1] = d[-1] / b[-1]
    for i in range(n-2, -1, -1):
        M[i] = (d[i] - c[i] * M[i+1]) / b[i]
    return M
```

```python
A, B, C, D = setup_tridiagonal_matrix(x, y)

print("Subdiagonal (A):", A)
print("Main diagonal (B):", B)
print("Superdiagonal (C):", C)
print("Right-hand side (D):", D)

M = thomas_algorithm(A, B, C, D)
M = np.insert(M, 0, 0)
M = np.append(M, 0)
print("M ",M)
```

```python
def evaluate_spline(M,x,y, xeval):
    n = len(x)
    h = np.diff(x)
    
    a = np.zeros(n)
    b = np.zeros(n)
    c = np.zeros(n)
    d = np.zeros(n)
    for i in range(0, n-1):
        a[i] = y[i]
        b[i] = ((y[i+1]-y[i])/h[i])-(h[i]/6)*(M[i+1]+2*M[i])
        c[i] = M[i]/2
        d[i] = (M[i+1]-M[i])/(6*h[i])
        

    start = np.searchsorted(x, xeval) - 1 # returns the interval index where wach value of xeval would be evaluated, so that each point gets the correct spline equation
    
    start[start<0] = 0
    start[start>(n-2)] = n-2
    # clips values so that anything below 0 = 0, anything above n-2 becomes n-2
    
    yeval = a[start] + b[start]*(xeval-x[start]) + c[start]*(xeval-x[start])**2 + d[start]*(xeval-x[start])**3
    # creates splines for each point
    return yeval
    
```

```python
# apply cubic spline and plot results
xeval = np.arange(1,57)
yeval = evaluate_spline(M,x,y,xeval)
plt.plot(xeval,yeval, 'r--', label = "Cubic Spline")
plt.scatter(xeval,yeval, label = "Interpolated Points")
plt.scatter(x,y, label = "Given Points")
plt.legend()
plt.xlabel("Quarter")
plt.ylabel("GDP Growth (%)")
plt.title("Quarter vs GDP Growth, Interpolated")

```

1. The system of equations being solved in the tridiagonal matrix is the second derivatives between each point, stored in a matrix M. This is because of the smoothness conditions of a natural cubic spline, specifically that the first and last second derivatives should be 0 (clamped cubic spline) and and the second derivatives from one segment to the next should be the same.

2. The interpolant is visually smooth, however, it is not guaranteed to be reasonable. Because cubic splines have to pass through every single given point, one outlier has a strong influence on the surrounding interpolated points. 

3. The Runge Phenomenon is the oscillatory behavior that occurs when fitting high degree polynomials to data, especially near the endpoints, causing the function to diverge rapidly instead of predicting the future trend of the data. When considering the merits of interpolation vs numerical stability when there are extreme outliers in the data, it's important to consider the context of the data. In this dataset, the spline is required to pass exactly through the 2020 Q2 value of −28.1%, which acts as a strong outlier. Because cubic splines are interpolating functions, they will exactly match this point, which can distort the shape of the spline in nearby intervals. If the data has local features, a polynomial or smoothing spline would not be able to accurately capture that. However, if the data is noisy, a cubic spline would pass through each point, not providing an accurate insight into the overall trend. Since the overall goal is to interpolate the data given, a spline would be more preferable due to the fact that it more accurately captures the local fluctuations between the points. However, if the goal was to predict a future quarter, a least squares model would provide the trend of the data, allowing predictions to be made.

### Part B - Polynomial & Least Squares Comparision

```python
# Create Vandermonde matrix
degree = 4
V = np.vander(x, N=degree+1, increasing=True)

# V.T * V
VT_V = np.dot(V.T, V)
# V.T * y
VT_y = np.dot(V.T, y)
# Solve for a (coefficients)

# vandermonde is O(n)

coefficients = np.linalg.solve(VT_V, VT_y)

# linalg.solve is O(n^3)

polynomial = np.poly1d(coefficients[::-1])

print(polynomial)

# Generate predicted y-values based on the fitted polynomial
y_pred = polynomial(x)
# plot results
plt.scatter(x, y, color='blue', label='Data points') # Original data points
plt.plot(x, y_pred, color='red', label=f'Polynomial fit of degree {degree}')
# Fitted curve
plt.title('Polynomial Fit')
plt.xlabel('Quarter')
plt.ylabel("GDP Growth (%)")
plt.legend()
plt.show()

ss_total = np.sum((y - np.mean(y))**2)
ss_res = np.sum((y - y_pred)**2)
r_squared = 1 - (ss_res / ss_total)
print(f'R-squared: {r_squared:.4f}')

mse = np.mean((y - y_pred)**2)
print(f'Mean Squared Error: {mse:.4f}')
```

```python
# to observe long term behavior, I extended the same polynomial from -30 to 120
xnew = np.arange(-30,120)
y_pred2 = polynomial(xnew)
plt.scatter(x, y, color='blue', label='Data points') # Original data points
plt.plot(xnew, y_pred2, color='red', label=f'Polynomial fit of degree {degree}')
plt.title('Polynomial Fit with extended quarters to show long-term trend')
plt.xlabel('Quarter')
plt.ylabel("GDP Growth (%)")
plt.legend()
```

```python
# plot cubic spline and degree 4 polyfit together
plt.plot(xeval,yeval, 'r', label = "Cubic Spline")
plt.scatter(x,y, label = "Given Points")
plt.plot(x, y_pred, color='orange', label=f'Polynomial fit of degree {degree}')
plt.xlabel("Quarter")
plt.ylabel("GDP Growth (%)")
plt.title("Quarter vs GDP Growth, Cubic Spline and Polyfit")
plt.legend()
plt.show()
```

1. As seen in the graph, the cubic spline is much better at reproducing specific data points due to the fact that it passes through every point and captures the local trend of the data. On the other hand, the degree-4 polynomial better captures the overall trend of the data. This is because it is not sensitive to noise. Interpolating data is better when the points are verified, while approximation is more robust but less accurate. When deciding whether to interpolate or approximate, it's important to consider how important that local features of the data are. This is because interpolation is extremely sensitive to noise.


```python
# Create a method for calculating slope and y-intercept for a least squares line using formula
def leastSquares(X,Y):
    sigX = np.sum(X)
    sigY = np.sum(Y)
    xSq = X**2
    sigX2 = np.sum(xSq)
    sigXY = np.dot(X,Y)
    n = len(X)
    
    m = (n*sigXY - sigX*sigY)/(n*sigX2 - sigX**2)
    c = (sigY - m*sigX)/n
    
    return m,c
```

```python
# excluding COVID outlier points
xnocovid = np.concatenate((x[:13], x[17:]))
ynocovid = np.concatenate((y[:13], y[17:]))


# calculate least squares linear model, plot it
LSm, LSc = leastSquares(xnocovid,ynocovid)
print(f"Least Squares Slope: {LSm}, Least Squares Intercept: {LSc}")

plt.scatter(xnocovid,ynocovid, label = "Given Points")
plt.plot(xnocovid,LSm*xnocovid+LSc,'r', label = "Least Squares Line")
plt.title('Least Squares Plot')
plt.xlabel("Quarter")
plt.ylabel("GDP Growth (%)")
plt.legend()
plt.show()

# residual plot
resid = ynocovid-(LSm*xnocovid+LSc)
plt.scatter(xnocovid,resid, label = "Residual")
plt.title('Least Squares Residual Plot')
plt.plot(xnocovid,np.zeros_like(ynocovid),'r--',label = "y = 0")
plt.xlabel("Quarter")
plt.ylabel("Residual (predicted y-actual y)")
plt.legend()
plt.show()

# r-squared and MSE for quantifying error
ss_total = np.sum((ynocovid - np.mean(ynocovid))**2)
ss_res = np.sum((ynocovid - (LSm*xnocovid+LSc))**2)
r_squared = 1 - (ss_res / ss_total)
print(f'R-squared: {r_squared:.4f}')

mse = np.mean((ynocovid - (LSm*xnocovid+LSc))**2)
print(f'Mean Squared Error: {mse:.4f}')
```

2. After excluding the COVID quarters, a least squares linear model yielded a slope of m=-0.001907, which represents the average change in GDP growth per quarter. While a linear trend can provide a rough approximation of long-term growth, it does not fully capture the cyclical and volatile nature of economic data. The model assumes a constant linear relationship, which is an assumption that is violated in real economic data, as seen by the excluded COVID quarters. The R-squared value is 0.0004, meaning 0.04% of the data can be explained by the least squares line, indicating that this data is not accurately captured by the least squares line. The Mean Squared Error of the data is 2.08, which is a relatively close fit. These indicators show that the Least Squares model has small errors, but does not capture meaningful variation in the data. However, it provides a long-term prediction that is more stable than the polynomial fitting, which appears to oscillate when extended. 

### Part C - Method Justification

For a policymaker who needs to estimate GDP growth between two known points, the method I would recommend would be a cubic spline. A cubic spline is constructed by creating a piecewise of cubic functions connecting each given point to the next while also keeping the function "smooth," meaning the second derivatives of the pieces connected by each point is the same. In this context, this means that the cubic spline would take the two known points and provide a cubic function that connects them. This makes the cubic spline more locally accurate as it only takes into account the two points that create the segment of the spline, while a polynomial fitting approach looked at between two points can be distorted by a large outlier later on in the data. The mean square error of the 4th degree polynomial is 100.6298, meaning that the average difference between the predicted point and actual point is very high. Cubic splines are the best for interpolation because they are locally accurate, while polynomial fitting is better for displaying trends of the data because it uses all of the data to create the model.


For the cubic spline algorithm, the system of equations is tridiagonal and can be solved in O(n) time using the Thomas algorithm, making it very efficient even for large datasets. For the polynomial fitting method, the time complexity is O(n^3) because of the LU decomposition behind np.linalg.solve, which is used to extract the coefficients from the Vandermonde matrix. The least squares implementation has a Big O of O(n) because the operations involved (sum, dot product) also have a time complexity of O(n). Because cubic splines scale linearly and remain stable while capturing local behavior, their Big O  reinforces the recommendation to use splines for interpolation. However, if the goal shifts toward modeling trends or handling noisy data, the higher computational cost of polynomial fit is justified by its improved numerical stability and robustness to outliers.

### Conclusion
Overall, cubic spline interpolation worked best for estimating values between data points because it creates a smooth curve and adapts to local changes without being too affected by extreme values. Polynomial fitting was less effective for interpolation since it is sensitive to outliers and can create unrealistic behavior across the entire dataset, which was reflected in its higher error. For trend analysis, the linear model performed better after removing the COVID quarters because it captures the general direction of GDP growth without trying to fit every fluctuation. This shows that different methods serve different purposes: splines are better for local accuracy, while simpler models like linear regression are more useful for understanding overall trends, and more complex global models can become unreliable when the data includes noise or extreme events.
